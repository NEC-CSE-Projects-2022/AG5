```python
import os
import cv2

# 🔁 Update this with your actual file name
folder_path = '/content/drive/MyDrive/archive/VideoDataset/Cycles/Cycle_1/'
video_file = 'Cycle_1_Clip.mp4'  # Replace with exact file name from step 2
video_path = '/content/drive/MyDrive/archive/VideoDataset/Cycles/Cycle_1/Cycle_1_task_0.mp4'

# Extract frames
def extract_frames(video_path):
    if not os.path.exists(video_path):
        print("❌ Error: Video path does not exist.")
        return []

    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        print("❌ Error: Cannot open video file.")
        return []

    frames = []
    while True:
        ret, frame = cap.read()
        if not ret:
            break
        frames.append(frame)

    cap.release()
    return frames

frames = extract_frames(video_path)
print(f"✅ Extracted {len(frames)} frames.")
```

```python
import mediapipe as mp
import numpy as np
import cv2

mp_hands = mp.solutions.hands
hands = mp_hands.Hands(static_image_mode=True, max_num_hands=1, min_detection_confidence=0.5)

skeleton_data = []

for i, frame in enumerate(frames):
    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    result = hands.process(rgb)

    if result.multi_hand_landmarks:
        print(f"✅ Hand detected in frame {i}")
        landmarks = result.multi_hand_landmarks[0].landmark
        coords = np.array([[lm.x, lm.y, lm.z] for lm in landmarks]).flatten()
    else:
        print(f"❌ No hand detected in frame {i}")
        coords = np.zeros(63)

    skeleton_data.append(coords)

skeleton_data = np.array(skeleton_data)
print("✅ Done! Skeleton shape:", skeleton_data.shape)
```

```python
import matplotlib.pyplot as plt
mp_drawing = mp.solutions.drawing_utils

# Show first frame with detection
rgb = cv2.cvtColor(frames[0], cv2.COLOR_BGR2RGB)
result = hands.process(rgb)

if result.multi_hand_landmarks:
    annotated = rgb.copy()
    mp_drawing.draw_landmarks(annotated, result.multi_hand_landmarks[0], mp_hands.HAND_CONNECTIONS)
    plt.imshow(annotated)
    plt.title("Hand Detected")
    plt.axis(False)
    plt.show()
else:
    print("❌ No hand detected in frame 0")
```

```python
import numpy as np
from tensorflow.keras.utils import to_categorical

# Dummy skeleton data: 100 samples, 219 frames, 63 features (21 hand joints × 3)
X = np.random.rand(100, 219, 63).astype(np.float32)
y = np.random.randint(0, 7, size=(100,))
y_categorical = to_categorical(y, num_classes=7)
```

```python
import numpy as np

# Compute velocity
velocity = np.diff(X, axis=1)
velocity = np.pad(velocity, ((0, 0), (0, 1), (0, 0)))  # Pad to match shape

# Concatenate original + velocity → shape becomes (N, 219, 126)
X_with_velocity = np.concatenate([X, velocity], axis=-1)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, Masking
from tensorflow.keras.optimizers import Adam

model = Sequential([
    Masking(mask_value=0.0, input_shape=(219, 63)),  # To ignore zero-padded frames
    LSTM(128, return_sequences=False),
    Dropout(0.5),
    Dense(64, activation='relu'),
    Dense(7, activation='softmax')  # 7 task classes
])

model.compile(optimizer=Adam(learning_rate=0.001),
              loss='categorical_crossentropy',
              metrics=['accuracy'])

model.summary()
```

```python
history = model.fit(
    X, y_categorical,
    validation_split=0.2,
    epochs=100,
    batch_size=8,
    shuffle=True
)
```

```python
model.save("lstm_model.h5")
print("✅ Model saved as 'lstm_model.h5'")
```

```python
# Plot accuracy
plt.plot(history.history['accuracy'], label='Train Acc')
plt.plot(history.history['val_accuracy'], label='Val Acc')
plt.title('Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.show()

# Plot loss
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Val Loss')
plt.title('Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.show()
```
